## Example 1: Basic InlineAgent Usage (FIXED)

Direct InlineAgent invocation for drug review collection with corrected OpenAPI schema:

```python
def invoke_inline_agent_direct(drug_name: str, max_results: int = 10):
    """
    CORRECTED Direct InlineAgent invocation example
    """
    session_id = f"pr-sentiment-{datetime.now().strftime('%Y%m%d-%H%M%S')}-{uuid.uuid4().hex[:8]}"
    
    # CORRECTED InlineAgent configuration with proper OpenAPI schema
    inline_agent_config = {
        'sessionId': session_id,
        'foundationModel': FOUNDATION_MODEL,
        'instruction': f'''You are a pharmaceutical data collection specialist that helps collect patient reviews for sentiment analysis.

Your task is to:
1. Use the TavilySearchProcessor function to search for patient reviews of "{drug_name}" on Drugs.com
2. Process and structure the collected review data
3. Provide a comprehensive summary of the data collection results
4. Handle any errors gracefully and provide clear status updates

Always maintain compliance with platform terms of service and data privacy guidelines.
Focus on collecting authentic patient experiences and reviews for sentiment analysis purposes.''',
        
        'actionGroups': [
            {
                'actionGroupName': 'TavilySearchProcessor',
                'description': 'Search for drug reviews on Drugs.com using Tavily API and process results',
                'actionGroupExecutor': {
                    'lambda': LAMBDA_FUNCTION_ARN
                },
                'apiSchema': {
                    'payload': json.dumps({
                        "openapi": "3.0.0",
                        "info": {
                            "title": "Tavily Search API", 
                            "version": "1.0.0"
                        },
                        "paths": {
                            "/search-drug-reviews": {
                                "post": {
                                    "description": "Search for drug reviews on Drugs.com using Tavily",
                                    "operationId": "search-drug-reviews",  # FIXED: Use hyphens to match apiPath
                                    "parameters": [
                                        {
                                            "name": "drug_name",
                                            "in": "query",
                                            "description": "Name of the drug to search for",
                                            "required": True,
                                            "schema": {"type": "string"}
                                        },
                                        {
                                            "name": "max_results",
                                            "in": "query", 
                                            "description": "Maximum number of results to return",
                                            "required": False,
                                            "schema": {"type": "integer", "default": 20}
                                        }
                                    ],
                                    # FIXED: Added missing responses section
                                    "responses": {
                                        "200": {
                                            "description": "Successful search results",
                                            "content": {
                                                "application/json": {
                                                    "schema": {
                                                        "type": "object",
                                                        "properties": {
                                                            "status": {"type": "string"},
                                                            "reviews": {"type": "array"},
                                                            "total_reviews_processed": {"type": "integer"}
                                                        }
                                                    }
                                                }
                                            }
                                        }
                                    }
                                }
                            }
                        }
                    })
                }
            }
        ],
        
        'inputText': f'Please search for patient reviews of "{drug_name}" on Drugs.com. Collect up to {max_results} reviews and provide a structured summary of the data collection results.',
        
        'enableTrace': True,
        'idleSessionTTLInSeconds': 3600
    }
    
    print(f"🚀 Invoking InlineAgent for {drug_name}...")
    
    try:
        response = bedrock_agent_runtime.invoke_inline_agent(**inline_agent_config)
        
        # Process streaming response
        collected_data = []
        trace_events = []
        
        for event in response['completion']:
            if 'chunk' in event:
                chunk = event['chunk']
                if 'bytes' in chunk:
                    text = chunk['bytes'].decode('utf-8')
                    collected_data.append(text)
                    print(f"📝 Received: {text[:100]}...")
                    
            elif 'trace' in event:
                trace_events.append(event['trace'])
                print(f"🔍 Trace event received")
        
        full_response = ''.join(collected_data)
        
        return {
            'status': 'success',
            'response': full_response,
            'trace_events': len(trace_events),
            'session_id': session_id
        }
        
    except Exception as e:
        print(f"❌ Error: {str(e)}")
        return {
            'status': 'error',
            'error': str(e)
        }

# Test with a sample drug
print("Testing CORRECTED direct InlineAgent invocation...")
result = invoke_inline_agent_direct("Lipitor", max_results=5)
print(f"\n✅ Result status: {result['status']}")
if result['status'] == 'success':
    print(f"📊 Response length: {len(result['response'])} characters")
    print(f"🔍 Trace events: {result['trace_events']}")
    print(f"📝 Response preview: {result['response'][:300]}...")
else:
    print(f"❌ Error: {result.get('error', 'Unknown error')}")
```

## Example 2: Using the Data Collector Class

Higher-level interface with built-in processing:

```python
# Load the PR Sentiment Data Collector
from pr_sentiment_data_collector import PRSentimentDataCollector, DataCollectionError

# Initialize the collector
collector = PRSentimentDataCollector(region_name=AWS_REGION)
print("✅ PR Sentiment Data Collector initialized")

# Test the data collector class
try:
    print("🧪 Testing PR Sentiment Data Collector...")
    
    # Collect reviews for a drug
    results = collector.collect_drug_reviews(
        drug_name="Metformin",
        max_results=10,
        lambda_function_arn=LAMBDA_FUNCTION_ARN
    )
    
    print(f"\n✅ Collection Status: {results['status']}")
    print(f"📊 Drug: {results['drug_name']}")
    print(f"📝 Response Length: {len(results.get('raw_response', ''))} characters")
    print(f"\n Raw Response: \n {results.get('raw_response','')}")
    # print json results
    print(f"\nJSON Results: \n {json.dumps(results, indent=2)}")
    
    # Check for structured data
    structured_data = results.get('structured_data')
    if structured_data:
        print(f"\n📋 Structured Data Found:")
        if 'reviews' in structured_data:
            print(f"   - Reviews: {len(structured_data['reviews'])}")
        if 'total_reviews_processed' in structured_data:
            print(f"   - Total Processed: {structured_data['total_reviews_processed']}")
    
    # Display metadata
    metadata = results.get('collection_metadata', {})
    print(f"\n🕒 Collection Time: {metadata.get('collection_time')}")
    print(f"🔍 Trace Events: {metadata.get('trace_events', 0)}")
    
except DataCollectionError as e:
    print(f"❌ Data Collection Error: {str(e)}")
except Exception as e:
    print(f"❌ Unexpected Error: {str(e)}")
```

## Example 3: Batch Processing Multiple Drugs

Process multiple drugs for comparative analysis:

```python
def batch_process_drugs(drug_list: List[str], max_results_per_drug: int = 5):
    """
    Process multiple drugs for comparative analysis
    """
    results = {}
    
    for drug in drug_list:
        print(f"\n🔄 Processing {drug}...")
        try:
            result = collector.collect_drug_reviews(
                drug_name=drug,
                max_results=max_results_per_drug,
                lambda_function_arn=LAMBDA_FUNCTION_ARN
            )
            
            results[drug] = result
            
            # Quick summary
            if result.get('structured_data') and 'reviews' in result['structured_data']:
                review_count = len(result['structured_data']['reviews'])
                print(f"   ✅ Collected {review_count} reviews for {drug}")
            else:
                print(f"   ⚠️ No structured reviews found for {drug}")
                
        except Exception as e:
            print(f"   ❌ Error processing {drug}: {str(e)}")
            results[drug] = {'status': 'error', 'error': str(e)}
    
    return results

# Example: Process multiple cholesterol medications
cholesterol_drugs = ["Lipitor", "Crestor"]

print("🧪 Batch processing cholesterol medications...")
batch_results = batch_process_drugs(cholesterol_drugs, max_results_per_drug=3)

# Summary of batch results
print("\n📊 Batch Processing Summary:")
for drug, result in batch_results.items():
    status = result.get('status', 'unknown')
    if status == 'completed':
        structured_data = result.get('structured_data', {})
        review_count = len(structured_data.get('reviews', []))
        print(f"   {drug}: ✅ {review_count} reviews collected")
    else:
        print(f"   {drug}: ❌ {status}")
```
